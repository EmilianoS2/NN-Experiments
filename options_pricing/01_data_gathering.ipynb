# 01 — Data gathering

We are not downloading historical prices yet. For this surrogate, we need many valid combinations of market conditions and contract terms. We will sample those conditions and later use Monte Carlo to create the target prices.

```python
from pathlib import Path
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n_scenarios = 2_000
n_assets = 5
assets = [f'spot_{i+1}' for i in range(n_assets)] + [f'vol_{i+1}' for i in range(n_assets)]
```

```python
# Market inputs and contract inputs. These ranges define the domain
# in which the trained MLP is intended to be reliable.
spots = rng.uniform(80.0, 120.0, size=(n_scenarios, n_assets))
vols = rng.uniform(0.10, 0.50, size=(n_scenarios, n_assets))
weights = np.full(n_assets, 1 / n_assets)
# The basket is an average of five stocks, so it is worth roughly $100.
strike = rng.uniform(80.0, 120.0, size=n_scenarios)
maturity = rng.uniform(0.25, 2.0, size=n_scenarios)
rate = rng.uniform(0.01, 0.05, size=n_scenarios)
correlation = rng.uniform(0.0, 0.80, size=n_scenarios)

data = pd.DataFrame(np.column_stack([spots, vols, strike, maturity, rate, correlation]),
                    columns=assets + ['strike', 'maturity', 'rate', 'correlation'])
data.head()
```

```python
output_dir = Path('data')
output_dir.mkdir(exist_ok=True)
data.to_csv(output_dir / 'raw_scenarios.csv', index=False)
print('saved', len(data), 'scenarios to', output_dir / 'raw_scenarios.csv')
```

# 02 — Data preparation

This notebook checks the sampled inputs, converts them to a numeric matrix, and creates the tensor-ready feature file. We do not normalize using future data: later, the training notebook will calculate normalization statistics from the training split only.

```python
from pathlib import Path
import numpy as np
import pandas as pd

data = pd.read_csv(Path('data') / 'raw_scenarios.csv')
data.info()
```

```python
# These are the conditions required for every input to be usable.
assert not data.isna().any().any()
assert (data.filter(like='spot_') > 0).all().all()
assert (data.filter(like='vol_') > 0).all().all()
assert (data['strike'] > 0).all()
assert (data['maturity'] > 0).all()
assert (data['rate'] > -1).all()
assert data['correlation'].between(0, 1).all()

feature_names = list(data.columns)
X = data.to_numpy(dtype=np.float32)
print('features:', feature_names)
print('shape:', X.shape)
```

```python
output_dir = Path('artifacts')
output_dir.mkdir(exist_ok=True)
np.save(output_dir / 'features.npy', X)
pd.Series(feature_names).to_json(output_dir / 'feature_names.json', index=False)
print(f'saved tensor-ready features to {output_dir}')
```
